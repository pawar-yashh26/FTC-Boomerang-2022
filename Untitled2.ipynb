```python
# STEP 1: Get the data — Tiny Shakespeare (~1MB of plays)
!wget -q -O input.txt https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt
text = open('input.txt', encoding='utf-8').read()
print("Characters in dataset:", len(text))
print(text[:300])
```

```python
# STEP 2 & 3: Tokenizer (characters -> numbers) and training examples
import torch

chars = sorted(set(text))
vocab_size = len(chars)
stoi = {c: i for i, c in enumerate(chars)}   # char -> number
itos = {i: c for c, i in stoi.items()}       # number -> char
encode = lambda s: [stoi[c] for c in s]
decode = lambda l: ''.join(itos[i] for i in l)
print("Vocab size:", vocab_size)
print(encode("hello"), "->", decode(encode("hello")))

# Turn the whole text into numbers; 90% train, 10% test
data = torch.tensor(encode(text), dtype=torch.long)
n = int(0.9 * len(data))
train_data, val_data = data[:n], data[n:]

batch_size = 32   # chunks of text per training step
block_size = 64   # how many characters the model can look back at
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print("Running on:", device)

def get_batch(split):
    d = train_data if split == 'train' else val_data
    ix = torch.randint(len(d) - block_size, (batch_size,))
    x = torch.stack([d[i:i+block_size] for i in ix])        # input characters
    y = torch.stack([d[i+1:i+block_size+1] for i in ix])    # answers = the NEXT character
    return x.to(device), y.to(device)

xb, yb = get_batch('train')
print("Input :", repr(decode(xb[0][:20].tolist())))
print("Target:", repr(decode(yb[0][:20].tolist())))
```

```python
# STEPS 6-8: The GPT model (self-attention + Transformer blocks)
import torch.nn as nn
from torch.nn import functional as F
torch.manual_seed(1337)

n_embd, n_head, n_layer, dropout = 128, 4, 4, 0.1

class Head(nn.Module):
    """One head of self-attention: each character looks back at the earlier ones."""
    def __init__(self, head_size):
        super().__init__()
        self.key   = nn.Linear(n_embd, head_size, bias=False)  # "what do I contain?"
        self.query = nn.Linear(n_embd, head_size, bias=False)  # "what am I looking for?"
        self.value = nn.Linear(n_embd, head_size, bias=False)  # "what do I share?"
        self.register_buffer('tril', torch.tril(torch.ones(block_size, block_size)))
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        B, T, C = x.shape
        k, q = self.key(x), self.query(x)
        wei = q @ k.transpose(-2, -1) * k.shape[-1] ** -0.5           # attention scores
        wei = wei.masked_fill(self.tril[:T, :T] == 0, float('-inf'))   # mask: no peeking ahead
        wei = self.dropout(F.softmax(wei, dim=-1))
        return wei @ self.value(x)

class MultiHeadAttention(nn.Module):
    """Several attention heads in parallel, then combined."""
    def __init__(self, num_heads, head_size):
        super().__init__()
        self.heads = nn.ModuleList([Head(head_size) for _ in range(num_heads)])
        self.proj = nn.Linear(n_embd, n_embd)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        out = torch.cat([h(x) for h in self.heads], dim=-1)
        return self.dropout(self.proj(out))

class FeedForward(nn.Module):
    """A small neural network applied to each position: lets it 'think' about what it gathered."""
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(n_embd, 4 * n_embd), nn.ReLU(),
            nn.Linear(4 * n_embd, n_embd), nn.Dropout(dropout))

    def forward(self, x):
        return self.net(x)

class Block(nn.Module):
    """Transformer block: attention (communicate) + feed-forward (compute), with residuals + layer norm."""
    def __init__(self):
        super().__init__()
        self.sa = MultiHeadAttention(n_head, n_embd // n_head)
        self.ffwd = FeedForward()
        self.ln1, self.ln2 = nn.LayerNorm(n_embd), nn.LayerNorm(n_embd)

    def forward(self, x):
        x = x + self.sa(self.ln1(x))     # residual connection
        x = x + self.ffwd(self.ln2(x))
        return x

class GPT(nn.Module):
    def __init__(self):
        super().__init__()
        self.token_emb = nn.Embedding(vocab_size, n_embd)     # what each character means
        self.pos_emb = nn.Embedding(block_size, n_embd)       # where it sits in the sequence
        self.blocks = nn.Sequential(*[Block() for _ in range(n_layer)])
        self.ln_f = nn.LayerNorm(n_embd)
        self.lm_head = nn.Linear(n_embd, vocab_size)          # scores for the next character

    def forward(self, idx, targets=None):
        B, T = idx.shape
        x = self.token_emb(idx) + self.pos_emb(torch.arange(T, device=idx.device))
        logits = self.lm_head(self.ln_f(self.blocks(x)))
        loss = None
        if targets is not None:
            loss = F.cross_entropy(logits.view(B * T, -1), targets.view(B * T))
        return logits, loss

    @torch.no_grad()
    def generate(self, idx, max_new_tokens):
        for _ in range(max_new_tokens):
            logits, _ = self(idx[:, -block_size:])
            probs = F.softmax(logits[:, -1, :], dim=-1)
            idx = torch.cat([idx, torch.multinomial(probs, 1)], dim=1)   # pick the next character
        return idx

model = GPT().to(device)
print(f"Model size: {sum(p.numel() for p in model.parameters())/1e6:.2f} million parameters")
print("Before training it writes gibberish:")
print(decode(model.generate(torch.zeros((1, 1), dtype=torch.long, device=device), 100)[0].tolist()))
```

```python
# STEP 5: Training loop — watch the loss go down
max_iters, eval_interval = 3000, 300
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3)

@torch.no_grad()
def estimate_loss():
    model.eval()
    out = {}
    for split in ['train', 'val']:
        losses = torch.zeros(50)
        for k in range(50):
            X, Y = get_batch(split)
            losses[k] = model(X, Y)[1].item()
        out[split] = losses.mean().item()
    model.train()
    return out

for step in range(max_iters + 1):
    if step % eval_interval == 0:
        l = estimate_loss()
        print(f"step {step:5d} | train loss {l['train']:.3f} | test loss {l['val']:.3f}")
    xb, yb = get_batch('train')
    logits, loss = model(xb, yb)
    optimizer.zero_grad()
    loss.backward()      # compute gradients
    optimizer.step()     # nudge the weights
```

```python
# Generate new "Shakespeare" from the trained model
start = torch.zeros((1, 1), dtype=torch.long, device=device)
print(decode(model.generate(start, max_new_tokens=600)[0].tolist()))
```
